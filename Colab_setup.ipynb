### Step 3: Create the Python 3.10 environment
Run this setup for each new Colab runtime. Training and testing explicitly use its Python executable; the notebook kernel is unchanged. No kernel restart is required.

```python
import subprocess
from pathlib import Path
import os
def run(args, stage=None):
    if stage:
        print(stage, flush=True)
    subprocess.run([str(arg) for arg in args], check=True)

root = Path('/content')
env = root / 'emgnotebook-env'
mamba = root / 'bin/micromamba'
if not mamba.exists():
    run(['bash', '-c', 'set -o pipefail; curl -fsSL https://micro.mamba.pm/api/micromamba/linux-64/latest | tar -xj -C /content bin/micromamba'])
if not (env / 'bin/python').exists():
    run([mamba, 'create', '-y', '-p', env, '-c', 'conda-forge', 'python=3.10', 'pip=24.0'], stage='Create Python environment')
python = env / 'bin/python'
version_check = 'import sys; assert sys.version_info[:2] == (3, 10), sys.version; print(sys.version); print(sys.executable)'
run([python, '-c', version_check], stage='Verify Python 3.10')
run(['bash', '-c', 'apt-get update -qq && apt-get install -y -qq build-essential cmake libboost-all-dev'], stage='Install native build dependencies')
run([python, '-m', 'pip', 'install', '-r', 'requirements.txt'], stage='Install project dependencies')
run([python, '-m', 'pip', 'install', '-e', '.'], stage='Install local project')

os.environ["MPLBACKEND"] = "Agg"
```
